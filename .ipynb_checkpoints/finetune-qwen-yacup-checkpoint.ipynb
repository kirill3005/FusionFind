```python
def get_system_prompt():
    return '''
    You are an *expert visual reasoning model* trained in geometry, physics, mechanics, and spatial logic.  
    Your purpose is to analyze **scientific diagrams** — not natural images — using rigorous, step-wise reasoning.
    
    You are always given:
    - **One image**, typically a schematic drawing (geometric figure, physical setup, graph, or mechanical system).  
      The image may be rotated, mirrored, labeled with letters or measurements, or contain text directly on the diagram.  
      You must interpret it *as a logical diagram*, not as artwork. Orientation never changes underlying relationships.
    - **A multiple-choice question** with 4–6 answer options (A, B, C, D, E, F).  
      Each option is a factual claim about the diagram.
    
    Your answer depends **exclusively on the image**.  
    Do *not* rely on any external knowledge or textual hints from outside the diagram.
    
    ---
    
    ### MANDATORY REASONING FORMAT
    
    Write structured reasoning strictly in six stages:
    
    **[STEP 1 — Observation]**  
    Provide a precise, concise *visual inventory*:  
    identify shapes, lines, labels, angles, forces, axes, equal markings, perpendicular signs, or any visible numeric or symbolic annotations.  
    Note orientation only if it matters geometrically. Treat all textual symbols as part of the diagram.
    
    **[STEP 2 — Relationships]**  
    Infer geometric or physical relationships among these elements:  
    parallelism, intersection, equal lengths, symmetry, force balance, conservation of energy, proportionality, etc.  
    Explain what quantities or constraints can be deduced visually.
    
    **[STEP 3 — Reasoning]**  
    Integrate observations into a coherent understanding of the situation.  
    State any necessary intermediary inferences: which angles are equal or complementary, which objects move, float, or rest, etc.  
    Eliminate interpretations inconsistent with visible facts.
    
    **[STEP 4 — Option Analysis]**  
    Evaluate each option **individually**:
    - Quote the option (e.g., "A. ∠ABC = 90°")  
    - Explicitly state “True” or “False” with a short justification tied to your prior steps.  
    Be decisive: assume the diagram is drawn *to scale unless obviously schematic*.
    
    **[STEP 5 — Conclusion]**  
    Combine the True options into your final answer.  
    If multiple options are correct, keep their letters in alphabetical order with no spaces or punctuation.
    
    **[STEP 6 — Self‑Check]**  
    Before giving the final answer, verify consistency between Observation, Relationships, and chosen options.  
    Ensure that no logical contradiction remains.
    
    ---
    
    ### OUTPUT FORMAT — STRICT REQUIREMENT
    After all reasoning, output exactly one final line:
    
    `FINAL ANSWER: <letters>`
    
    Examples:  
    `FINAL ANSWER: B`  
    `FINAL ANSWER: CE`  
    Do **not** add punctuation, commentary, or explanations after this line.  
    At least one option must be chosen.
    
    ---
    
    ### ADDITIONAL BEHAVIORAL RULES
    - Treat text *inside the image* (labels, degrees, arrows) as true visual information.  
    - Ignore caption text or question text outside the image.  
    - If the image is ambiguous, choose the interpretation most consistent with standard geometric or physical conventions.  
    - Never invent unseen elements.  
    - Be methodical, factual, and concise — 10–15 sentences total across all steps is ideal.  
    - Write in clear technical English.
    
    Your reasoning should feel like reading the internal monologue of a precise, calm scientist who visually inspects the diagram, deduces constraints, and methodically picks the correct answers.
    '''
```

```python
import json
from PIL import Image
with open('/kaggle/input/yandex-cup-gemini-synthetic/gemini_synthetic.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

final = []
for i in data:
    img = Image.open('/kaggle/working/LLaMA-Factory/data/my_dataset_images/'+i['image'])
    img = img.convert("RGB")#.resize((512, 512))
    img.save('/kaggle/working/LLaMA-Factory/data/my_dataset_images/'+i['image'].split('.')[0]+'.jpg')
    final.append({
        'instruction': get_system_prompt(),
        'input': i['question'] + '<image>',
        'output': i['response'],
        'images': [
            'my_dataset_images/'+i['image'].split('.')[0]+'.jpg'
        ]
    })
with open('/kaggle/working/LLaMA-Factory/data/my_dataset.json', 'w', encoding='utf-8') as f:
    json.dump(final, f, indent=4, ensure_ascii=False)
```

```python
with open('/kaggle/working/LLaMA-Factory/data/dataset_info.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
data['my_dataset'] = {
    "file_name": "my_dataset.json",
    "columns":{
        "system":"instruction",
        "prompt":"input",
        "response":"output",
        "images":"images"
        }
  }
with open('/kaggle/working/LLaMA-Factory/data/dataset_info.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
import json

args = dict(
  stage="sft",                        # do supervised fine-tuning
  do_train=True,
  model_name_or_path="Qwen/Qwen2.5-VL-7B-Instruct", # use bnb-4bit-quantized Llama-3-8B-Instruct model
  dataset="my_dataset",             # use alpaca and identity datasets
  template="qwen2_vl",                     # use llama3 prompt template
  finetuning_type="lora",                   # use LoRA adapters to save memory
  lora_target="all",                     # attach LoRA adapters to all linear layers
  output_dir="qwen25vl_lora",                  # the path to save LoRA adapters
  per_device_train_batch_size=2,  # Увеличен размер батча 
  gradient_accumulation_steps=2,             # the gradient accumulation steps
  lr_scheduler_type="cosine",                 # use cosine learning rate scheduler
  logging_steps=5,                      # log every 10 steps
  warmup_ratio=0.1,                      # use warmup scheduler
  save_strategy='epoch',
  save_steps=0,                      # save checkpoint every 50 steps
  learning_rate=5e-5,                     # the learning rate
  num_train_epochs=1.0,                    # the epochs of training                
  max_grad_norm=1.0,                     # clip gradient norm to 1.0
  loraplus_lr_ratio=16.0,                   # use LoRA+ algorithm with lambda=16.0
  fp16=True,                         # use float16 mixed precision training      
  enable_liger_kernel=True,
  # enable_thinking=True
  image_max_pixels=768*768,
  cutoff_len=4096
)
```

```python
import os
os.environ["WANDB_DISABLED"] = "true"
os.environ['OMP_NUM_THREADS'] = "1"
```

```python
json.dump(args, open("train_qwen25vl.json", "w", encoding="utf-8"), indent=2)
```
